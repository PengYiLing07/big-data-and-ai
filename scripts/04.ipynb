```python
#   函数
import random
def 取名(姓=None,名=None ):
    ##   如果为空，则从类似 欧阳+吹雪 这样的组合中，随机选择一个姓和名  
    if 姓==None:
        姓=random.choice(['欧阳','司马','上官','诸葛','东方','独孤','南宫','西门'])
    if 名==None:
        名=random.choice(['吹雪','子涵','雨萱','浩然','欣怡','睿轩','梓涵','欣然'])
    return 姓 + 名  
```

```python
#   类和对象

##   类的定义
class Person:
    def __init__(self, 姓名, 年龄):
        self.姓名 = 姓名
        self.年龄 = 年龄

    def 介绍(self):
        print(f"你好，我是{self.姓名}，今年{self.年龄}岁。")
```

```python
王小利 = Person("王小利", 25)
王小利.介绍()  # 输出: 你好，我是王小利
```
